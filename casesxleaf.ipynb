# Experimenatció de nombre de nodes per fulla

```python
from miniproto1 import *
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def exp_retain(casexfulla):
    num_cases = 1000
    df_problemes = pd.read_csv('./data/problems8000.csv')
    rs = createRS(from_model= False, casebase=df_problemes, maxcasesxleaf=casexfulla)

    avaluacions, qualitat = [], []

    for _ in range(num_cases):
        newp_str = generate_new_problem()
        newp = eval(newp_str)
        newcas = Cas(id=(rs.indexacio.ultim_id),descripcio=rs.scaler.transform([newp.array])[0])    
        
        most_similar_cases = rs.retrieve(newp) # Retrieve

        distancies, casos, paths = zip(*most_similar_cases) # Reuse
        cas_dists = list(zip(distancies, casos))
        dicc_most_sim = rs.reuse(cas_dists, newcas)

        rs.revise(newcas, casos, dicc_most_sim, False) # Revise

        rs.retain(newcas, cas_dists, paths[0]) # Retain

        avaluacions.append(rs.evaluation())
        qualitat.append(rs.NMAE())
    return avaluacions, qualitat
```

```python
def plot_data(data, title, thresholds):
    for i, sublist in enumerate(data):
        plt.plot(sublist, label=f'Threshold {thresholds[i]}')

    plt.title(title)
    plt.xlabel('Índex de cas')
    plt.ylabel('Valor')
    plt.legend()
    plt.show()
```

```python
num_casos = [6, 8, 10, 12, 15, 20] 
avaluacions, nmaes = [], []

for casos in num_casos:
    aval, qual = exp_retain(casos)
    avaluacions.append(aval)
    nmaes.append(qual)


plot_data(avaluacions, 'Avaluacions per a diferents valoors de CasosxNode', num_casos)
plot_data(nmaes, 'Resultats de MNAE per a diferents valors de CasosxNode', num_casos)
```
