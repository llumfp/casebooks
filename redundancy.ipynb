# Experimentació amb la redundancia dels casos

```python
from miniproto1 import *
import pandas as pd
import matplotlib.pyplot as plt
import itertools
```

```python
"""
Per a cada valor de threshold, es genera un problema, i s'afegeixen 1000 casos nous.
Per a cada cas nou es mira:
- Accurcy del model rs.nmae(self)
- Avauluació -> rs.evaluation(self)
Finalment, es mira per a cada valor de thresholds quins són els millors resultats.
La maxcases leaf esta fixada a 10."""


def exp_initial_cases(parametres):
    num_cases = 1000
    df_problemes = pd.read_csv('./data/problems8000.csv')
    rs = createRS(from_model= False, casebase=df_problemes, params=parametres)

    avaluacio, qualitat = [], []

    for _ in range(num_cases):
        newp_str = generate_new_problem()
        newp = eval(newp_str)
        newcas = Cas(id=(rs.indexacio.ultim_id),descripcio=rs.scaler.transform([newp.array])[0])    
        
        most_similar_cases = rs.retrieve(newp) # Retrieve

        distancies, casos, paths = zip(*most_similar_cases) # Reuse
        cas_dists = list(zip(distancies, casos))
        dicc_most_sim = rs.reuse(cas_dists, newcas)

        rs.revise(newcas, casos, dicc_most_sim, False) # Revise

        rs.retain(newcas, cas_dists, paths[0]) # Retain

        avaluacio.append(rs.evaluation())
        qualitat.append(rs.NMAE())
    return avaluacio, qualitat
```

## Experimentació de la distància en la considera que un cas nou és red

```python
def exp_retain(threshold):
    num_cases = 1000
    df_problemes = pd.read_csv('./data/problems8000.csv')
    rs = createRS(from_model= False, casebase=df_problemes)

    avaluacio, qualitat = [], []

    for _ in range(num_cases):
        newp_str = generate_new_problem()
        newp = eval(newp_str)
        newcas = Cas(id=(rs.indexacio.ultim_id),descripcio=rs.scaler.transform([newp.array])[0])    
        
        most_similar_cases = rs.retrieve(newp) # Retrieve

        distancies, casos, paths = zip(*most_similar_cases) # Reuse
        cas_dists = list(zip(distancies, casos))
        dicc_most_sim = rs.reuse(cas_dists, newcas)

        rs.revise(newcas, casos, dicc_most_sim, False) # Revise

        rs.retain(newcas, cas_dists, paths[0], threshold) # Retain

        avaluacio.append(rs.evaluation)
        qualitat.append(rs.NMAE)
    return avaluacio, qualitat
```

```python
def plot_data(data, title, thresholds):
    for i, sublist in enumerate(data):
        plt.plot(sublist, label=f'Threshold {thresholds[i]}')

    plt.title(title)
    plt.xlabel('Índex de cas')
    plt.ylabel('Valor')
    plt.legend()
    plt.show()
```

```python
distance_thresholds = [0.7, 0.8, 0.9, 1.0, 1.1, 1.2, 1.3, 1.4, 1.5] 
avaluacions, nmaes = [], []

for threshold in distance_thresholds:
    aval, qual = exp_retain(threshold)
    avaluacions.append(aval)
    nmaes.append(qual)


plot_data(avaluacions, 'Avaluacions per a diferents valors de Thresholds', distance_thresholds)
plot_data(nmaes, 'Resultats de MNAE per a diferents valors de Thresholds', distance_thresholds)
```
